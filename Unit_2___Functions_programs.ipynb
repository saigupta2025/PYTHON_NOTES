```python
# develop a program to define a function with multiple return values
# Function with multiple return values
def calculate(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b
    return addition, subtraction, multiplication

# Main program
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

add, sub, mul = calculate(x, y)  # unpacking multiple return values

print("Addition:", add)
print("Subtraction:", sub)
print("Multiplication:", mul)
```

```python
# Function to find length of a string without using len()
def string_length(s):
    count = 0
    for char in s:   # loop through each character
        count += 1
    return count

# Main program
text = input("Enter a string: ")
length = string_length(text)

print("The length of the string is:", length)
```

```python
# Function with default arguments
def calculate_bill(item_name, price, quantity=1, tax_rate=0.05, discount=0.0):
    """
    Calculate final bill for an item
    :param item_name: Name of the item
    :param price: Price of one unit
    :param quantity: Quantity (default 1)
    :param tax_rate: Tax percentage (default 5%)
    :param discount: Discount percentage (default 0%)
    :return: Final amount after tax and discount
    """
    subtotal = price * quantity
    tax = subtotal * tax_rate
    discount_amount = subtotal * discount
    final_amount = subtotal + tax - discount_amount

    print(f"\nItem: {item_name}")
    print(f"Price per unit: {price}")
    print(f"Quantity: {quantity}")
    print(f"Subtotal: {subtotal}")
    print(f"Tax ({tax_rate*100}%): {tax}")
    print(f"Discount ({discount*100}%): {discount_amount}")
    print(f"Final Bill: {final_amount}")

    return final_amount


# Main program
print("----- Shopping Bill System -----")

# Using default quantity=1, tax=5%, discount=0%
calculate_bill("Notebook", 50)

# Providing custom quantity but using default tax and discount
calculate_bill("Pen", 10, quantity=5)

# Providing custom values for all
calculate_bill("Headphones", 1000, quantity=2, tax_rate=0.18, discount=0.10)
```

```python
# Function to check if substring exists in main string
def check_substring(main, sub):
    if sub in main:
        return True
    else:
        return False

# Main program
main = input("Enter the main string: ")
sub = input("Enter the substring to check: ")

if check_substring(main, sub):
    print(f"Yes The substring '{sub}' is present in '{main}'")
else:
    print(f"No  The substring '{sub}' is not present in '{main}'")
```
