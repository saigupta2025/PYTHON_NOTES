```python
# display a program to create, display, append, insert and reverse the order of the items in an array

from array import *      # “Import everything (all classes and functions) from the array module.”

# 1️ Create an array of integers
numbers = array('i', [10, 20, 30, 40, 50])      # array('i', [...])-->'i' means integer type
print("Original array:", numbers.tolist())   # tolist --> Converts array to normal list

# 2️ Display each element
print("\nArray elements:")
for num in numbers:
    print(num, end=" ")

# 3️ Append a new item
numbers.append(60)
print("\n\nAfter appending 60:", numbers.tolist())

# 4️ Insert an item at a specific position
numbers.insert(2, 25)  # insert 25 at index 2
print("After inserting 25 at position 2:", numbers.tolist())

# 5️ Reverse the order of items
numbers.reverse()
print("After reversing:", numbers.tolist())
```

```python
from abc import ABC, abstractmethod   # for abstraction

# 1️ ABSTRACTION — Base class (idea only, no direct use)
class Vehicle(ABC):
    def __init__(self, brand):
        self._brand = brand   # protected (Encapsulation: not fully public)

    @abstractmethod
    def fuel_type(self):      # abstract method — must be defined by child
        pass

    def show_brand(self):
        print(f"Brand: {self._brand}")


# 2️ INHERITANCE — Car and Bike inherit from Vehicle
class Car(Vehicle):
    def __init__(self, brand, color):
        super().__init__(brand)
        self.__color = color   # private variable (Encapsulation)

    def fuel_type(self):       # implementing abstract method
        return "Petrol or Diesel"

    def show_info(self):
        print(f"This is a {self.__color} car made by {self._brand}.")


class Bike(Vehicle):
    def fuel_type(self):       # implementing abstract method
        return "Petrol"

    def sound(self):           # specific behavior
        print("Bike says Vrooom! 🏍️")


# 3️ POLYMORPHISM — Same method name, different behavior
def vehicle_sound(vehicle):
    vehicle.show_brand()
    print("Fuel Type:", vehicle.fuel_type())


# 4️ OBJECT CREATION — using the classes
car1 = Car("Toyota", "Red")
bike1 = Bike("Yamaha")

# 5️ CALLING METHODS
car1.show_info()             # Object of Car class
bike1.sound()                # Object of Bike class

# Polymorphism in action:
print("\n--- Vehicle Sounds ---")
for v in [car1, bike1]:
    vehicle_sound(v)
```

```python
# We created a “Vehicle” idea (abstract class),
# built “Car” and “Bike” using that idea (inheritance),
# hid some details like brand and color (encapsulation),
# used the same method fuel_type() in different ways (polymorphism),
# and finally created real objects (car1, bike1).
```
