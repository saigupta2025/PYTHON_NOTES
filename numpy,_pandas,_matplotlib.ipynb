```python
import numpy as np
a = np.array([[1,2],[3,4]])
print(a)
b = np.array([[5,6],[7,8]])
print(b)
print("sum of two matrices::", a+b)

print("Transpose of a:", a.T)

print("product of a,b:", np.dot(a,b))

a_inv = np.linalg.inv(a)
print("Inverse of 2x2 Matrix:\n", a_inv)

a_det= np.linalg.det(a)
print("det of 2x2 Matrix:\n", a_det)

print("eigen values of a matrix :\n", np.linalg.eig(a))

np.linalg.matrix_rank(a)
```

## **Meaning of linalg**

# linalg stands for "Linear Algebra".

It is a submodule in NumPy —
# so np.linalg = NumPy Linear Algebra module.

```python
import pandas as pd

# Create a dictionary of data
data = {
    'Name': ['Aarav', 'Isha', 'Rohan', 'Sneha', 'Aditi', 'Vikram', 'Tara', 'Arjun', 'Nisha', 'Kiran'],
    'Math': [88, 92, 79, 93, 85, 72, 95, 89, 76, 84],
    'Science': [90, 85, 88, 91, 80, 70, 92, 83, 78, 86],
    'English': [85, 87, 82, 89, 84, 75, 93, 88, 79, 90],
    'City': ['Delhi', 'Mumbai', 'Pune', 'Chennai', 'Kolkata', 'Delhi', 'Pune', 'Chennai', 'Mumbai', 'Kolkata']
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print("---- Student Marks Data ----\n")
print(df)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
product_A = [12000, 15000, 17000, 13000, 18000, 20000]
product_B = [10000, 12000, 14000, 11000, 16000, 17500]
product_C = [8000, 9500, 12000, 10000, 13000, 14500]
```

```python
plt.figure(figsize=(6,3))
plt.plot(months, product_A, marker='o', label='Product A')
plt.plot(months, product_B, marker='s', label='Product B')
plt.plot(months, product_C, marker='^', label='Product C')

plt.title("Monthly Sales Trend", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
x = np.arange(len(months))
width = 0.25

plt.figure(figsize=(6,3))
plt.bar(x - width, product_A, width, label='Product A')
plt.bar(x, product_B, width, label='Product B')
plt.bar(x + width, product_C, width, label='Product C')

plt.title("Product-wise Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.xticks(x, months)
plt.legend()
plt.show()
```

```python
total_sales = [sum(product_A), sum(product_B), sum(product_C)]
products = ['Product A', 'Product B', 'Product C']

plt.figure(figsize=(3,3))
plt.pie(total_sales, labels=products, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title("Market Share by Product")
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(5,3))
plt.scatter(product_A, product_B, color='purple', s=100)
plt.title("Correlation between Product A and B Sales")
plt.xlabel("Product A Sales")
plt.ylabel("Product B Sales")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
all_sales = product_A + product_B + product_C

plt.figure(figsize=(5,3))
plt.hist(all_sales, bins=6, color='skyblue', edgecolor='blue')
plt.title("Sales Distribution")
plt.xlabel("Sales Amount ($)")
plt.ylabel("Frequency")
plt.grid(axis='y', alpha=0.3)
plt.show()
```
