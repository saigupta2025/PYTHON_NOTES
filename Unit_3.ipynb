```python
# Initial list
numbers = [10, 20, 30, 40, 50,60,70]
print("Original List:", numbers)

# 1) Addition (append new element at the end)
numbers.append(80)
print("After Addition (append 80):", numbers)

# 2) Insertion (insert element at specific position)
numbers.insert(2, 25)   # Insert 25 at index 2
print("After Insertion (25 at index 2):", numbers)

# 3) Slicing (extract part of the list)
sliced_list = numbers[1:4]   # from index 1 to 3
print("Sliced List (index 1 to 3):", sliced_list)
```

```python
# Sample list
numbers = [10, 25, 7, 42, 18, 66, 36, 44]

print("Original List:", numbers)

# 1) len():: Get length of list
print("Length of list:", len(numbers))

# 2) max() :: Get maximum element
print("Maximum value:", max(numbers))

# 3) min() :: Get minimum element
print("Minimum value:", min(numbers))

# 4) sum()::  Sum of all elements
print("Sum of all elements:", sum(numbers))

# 5) sorted():: Return sorted list
print("Sorted List:", sorted(numbers))
```
