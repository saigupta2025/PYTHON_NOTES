```python
# display a program to create, display, append, insert and reverse the order of the items in an array

from array import *      # “Import everything (all classes and functions) from the array module.”

# 1️ Create an array of integers
numbers = array('i', [10, 20, 30, 40, 50])      # array('i', [...])-->'i' means integer type
print("Original array:", numbers.tolist())   # tolist --> Converts array to normal list

# 2️ Display each element
print("\nArray elements:")
for num in numbers:
    print(num, end=" ")

# 3️ Append a new item
numbers.append(60)
print("\n\nAfter appending 60:", numbers.tolist())

# 4️ Insert an item at a specific position
numbers.insert(2, 25)  # insert 25 at index 2
print("After inserting 25 at position 2:", numbers.tolist())

# 5️ Reverse the order of items
numbers.reverse()
print("After reversing:", numbers.tolist())
```

```python
class Student:
    def __init__(self, name, age):
        # This is the constructor
        self.name = name   # Assign value to instance variable
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

# When you create the object, constructor runs automatically
s1 = Student("rama", 25)
s2 = Student("sita", 22)

s1.display()
s2.display()
```

```python
from abc import ABC, abstractmethod
import math

# Abstract base class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


# Circle subclass
class Circle(Shape):
    def __init__(self, radius):
        self.__radius = radius  # Encapsulation (private variable)

    def area(self):
        return math.pi * (self.__radius ** 2)

    def perimeter(self):
        return 2 * math.pi * self.__radius


# Square subclass
class Square(Shape):
    def __init__(self, side):
        self.__side = side

    def area(self):
        return self.__side ** 2

    def perimeter(self):
        return 4 * self.__side


# Triangle subclass
class Triangle(Shape):
    def __init__(self, a, b, c):
        self.__a = a
        self.__b = b
        self.__c = c

    def area(self):
        # Using Heron's formula
        s = (self.__a + self.__b + self.__c) / 2
        return math.sqrt(s * (s - self.__a) * (s - self.__b) * (s - self.__c))

    def perimeter(self):
        return self.__a + self.__b + self.__c


# ---- Main Program ----
def main():
    print("🏗️ Shape Calculation Program 🏗️")

    # Create objects of different shapes
    circle = Circle(5)
    square = Square(4)
    triangle = Triangle(3, 4, 5)

    # Display results
    print("\n🔵 Circle:")
    print(f"Area: {circle.area():.2f}")
    print(f"Perimeter: {circle.perimeter():.2f}")

    print("\n🟩 Square:")
    print(f"Area: {square.area():.2f}")
    print(f"Perimeter: {square.perimeter():.2f}")

    print("\n🔺 Triangle:")
    print(f"Area: {triangle.area():.2f}")
    print(f"Perimeter: {triangle.perimeter():.2f}")


if __name__ == "__main__":
    main()
```
