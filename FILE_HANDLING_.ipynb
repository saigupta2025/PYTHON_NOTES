```python
f= open("cse_class.txt", "w")
print("file name ::", f.name)
print("file mode ::", f.mode)
print("is file readable ::", f.readable())
print("is file writable ::", f.writable())
print("is file closed ::", f.closed)
f.close()
print("is file closed ::", f.closed)
```

```python
# using 'with' statement
with open("example.txt", "w") as file:
    file.write("Python File Handling Example\n")
    file.write("This file is handled using 'with' statement.\n")

# file is automatically closed after 'with' block
print(file.closed)   # Output: True
```

```python
with open("example.txt", "r") as file:
    print("Cursor position at start:", file.tell())   # 0

    data = file.read(10)  # read 10 characters
    print("Data read:", data)
    print("Cursor position now:", file.tell())        # after reading 10 characters
```

```python
with open("example.txt", "r") as file:
    print("Original position:", file.tell())
    file.seek(7)  # move cursor to 7th byte
    print("After seek(7), position:", file.tell())

    data = file.read(10)
    print("Data after seek:", data)
```
